# Dataset path retrival

```python
train_path = "train"
valid_path = "valid"
test_path = "test"
```

# Importing Libraries

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
```

# Data Preprocessing

### Training Image Preprocessing

```python
training_set = tf.keras.utils.image_dataset_from_directory(
    train_path,
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
    pad_to_aspect_ratio=False,
    data_format=None,
    format="tf",
    verbose=True,
)
```

### Validation Image Preprocessing

```python
validation_set = tf.keras.utils.image_dataset_from_directory(
    valid_path,
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
    pad_to_aspect_ratio=False,
    data_format=None,
    format="tf",
    verbose=True,
)
```

```python
training_set
```

```python
for x,y in training_set:
  print(x,x.shape)
  print(y,y.shape)
  break
```

## Building Model

```python
#To avoid Overshooting
#1. Choose small learning rate default 0.001 we are taking 0.0001
#2.There may be a chance of underfitting,so increase number of neuron
#3. Add more Convolutional layer to extract more feature from images ther may be possibility that model unable to capture relevant feature or model is confusing
```

```python
from tensorflow.keras.layers import Dense,Conv2D,MaxPool2D,Flatten,Dropout,BatchNormalization
from tensorflow.keras.models import Sequential
```

```python
model = Sequential()
```

```python
## Building Convolution Layer
```

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(256, 256, 3)),

    tf.keras.layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    tf.keras.layers.MaxPooling2D(2, 2),

    tf.keras.layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    tf.keras.layers.MaxPooling2D(2, 2),

    tf.keras.layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    tf.keras.layers.MaxPooling2D(2, 2),

    tf.keras.layers.Conv2D(256, (3, 3), activation='relu', padding='same'),
    tf.keras.layers.MaxPooling2D(2, 2),

    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dense(256, activation='relu'),
    tf.keras.layers.Dropout(0.5),

    tf.keras.layers.Dense(38, activation='softmax')
])
```

```python
model.add(Dropout(0.25))
```

```python
model.add(Flatten())
```

```python
model.add(Dense(units=1500,activation='relu'))
```

```python
model.add(Dropout(0.4))
```

```python
#Output layer
model.add(Dense(units=38,activation='softmax'))
```

### Compiling Model

```python
model.compile(optimizer=tf.keras.optimizers.Adam(
    learning_rate=0.0001
),loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

## Model Training

```python
training_history = model.fit(x=training_set,validation_data=validation_set,epochs=10)
```

## Model Evaluation

```python
# Model evaluation on training set
train_loss,train_acc = model.evaluate(training_set)
```

```python
print("Training Loss:",train_loss)
print("Training Accuracy:",train_acc)
```

```python
# Model evaluation on validation set
val_loss,val_acc = model.evaluate(validation_set)
```

```python
print("Validation Loss:",val_loss)
print("Validation Accuracy:",val_acc)
```

## Saving Model

```python
model.save("trained_model.keras")
```

```python
epochs = [i for i in range(1,11)]
plt.plot(epochs,training_history.history['accuracy'],color='red',label='Training Accuracy')
plt.plot(epochs,training_history.history['val_accuracy'],color='blue',label='Validation Accuracy')
plt.xlabel("No of Epochs")
plt.ylabel("Accuracy Result")
plt.title("Visualisation Accuracy Result")
plt.legend()
plt.show()
```
