## 2. Dataset paths

````python
train_path = "train"
valid_path = "valid"
test_path = "test"

# If your folders are inside another folder, write the full path.
# Example: train_path = "/content/PlantVillage/train"
````

## 3. Importing libraries

````python
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import json
import os

from sklearn.metrics import classification_report, confusion_matrix

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 20

print("TensorFlow version:", tf.__version__)
````

## 4. Checking the dataset folders

````python
# Check that all three folders exist
for path in [train_path, valid_path, test_path]:
    if not os.path.isdir(path):
        raise FileNotFoundError("Folder not found: " + path)

# Check that training, validation and test have the same class folders
train_classes = sorted([name for name in os.listdir(train_path)
                        if os.path.isdir(os.path.join(train_path, name))])
valid_classes = sorted([name for name in os.listdir(valid_path)
                        if os.path.isdir(os.path.join(valid_path, name))])
test_classes = sorted([name for name in os.listdir(test_path)
                       if os.path.isdir(os.path.join(test_path, name))])

if train_classes != valid_classes or train_classes != test_classes:
    raise ValueError("Class folders must be identical in train, valid and test.")

class_name = train_classes
num_classes = len(class_name)

print("Number of classes:", num_classes)
print("Classes:", class_name)
````

## 5. Data preprocessing

### Training images

````python
training_set = tf.keras.utils.image_dataset_from_directory(
    train_path,
    labels="inferred",
    label_mode="int",
    class_names=class_name,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=42
)
````

### Validation images

````python
validation_set = tf.keras.utils.image_dataset_from_directory(
    valid_path,
    labels="inferred",
    label_mode="int",
    class_names=class_name,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)
````

### Test images

````python
test_set = tf.keras.utils.image_dataset_from_directory(
    test_path,
    labels="inferred",
    label_mode="int",
    class_names=class_name,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

# Load data faster during model training
AUTOTUNE = tf.data.AUTOTUNE
training_set = training_set.prefetch(AUTOTUNE)
validation_set = validation_set.prefetch(AUTOTUNE)
test_set = test_set.prefetch(AUTOTUNE)
````

## 6. Display some training images

````python
plt.figure(figsize=(10, 8))

for images, labels in training_set.take(1):
    for i in range(min(9, len(images))):
        plt.subplot(3, 3, i + 1)
        plt.imshow(images[i].numpy().astype("uint8"))
        plt.title(class_name[int(labels[i])])
        plt.axis("off")

plt.tight_layout()
plt.show()
````

## 7. Data augmentation

````python
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1)
])
````

## 8. Building the model

````python
# Use MobileNetV2 with ImageNet pretrained weights
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

# First train only the new classification layers
base_model.trainable = False

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(224, 224, 3)),
    data_augmentation,
    tf.keras.layers.Lambda(
        tf.keras.applications.mobilenet_v2.preprocess_input
    ),
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(num_classes, activation="softmax")
])

model.summary()
````

## 9. Compiling the model

````python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
````

## 10. Model training

````python
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=0.0000001
    ),
    tf.keras.callbacks.ModelCheckpoint(
        "best_plant_disease_model.keras",
        monitor="val_loss",
        save_best_only=True
    )
]

training_history = model.fit(
    training_set,
    validation_data=validation_set,
    epochs=EPOCHS,
    callbacks=callbacks
)
````

## 11. Fine-tuning the model (extra improvement)

````python
# Unfreeze only the last 30 layers of MobileNetV2
base_model.trainable = True

for layer in base_model.layers[:-30]:
    layer.trainable = False

# Keep Batch Normalization layers frozen
for layer in base_model.layers:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.00001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

fine_tune_history = model.fit(
    training_set,
    validation_data=validation_set,
    epochs=10,
    callbacks=callbacks
)
````

## 12. Model evaluation

````python
# Load the best model selected using validation loss
model = tf.keras.models.load_model("best_plant_disease_model.keras")

train_loss, train_acc = model.evaluate(training_set)
print("Training Loss:", train_loss)
print("Training Accuracy:", train_acc)

val_loss, val_acc = model.evaluate(validation_set)
print("Validation Loss:", val_loss)
print("Validation Accuracy:", val_acc)

test_loss, test_acc = model.evaluate(test_set)
print("Test Loss:", test_loss)
print("Test Accuracy:", test_acc)
````

## 13. Saving the model

````python
model.save("trained_plant_disease_model.keras")

with open("class_names.json", "w") as f:
    json.dump(class_name, f)

print("Model and class names saved successfully.")
````

## 14. Recording training history

````python
# Combine the history from both training stages
history = {}

for key in training_history.history:
    history[key] = training_history.history[key] + fine_tune_history.history[key]

with open("training_hist.json", "w") as f:
    json.dump(history, f)
````

## 15. Accuracy visualization

````python
epochs_range = range(1, len(history["accuracy"]) + 1)

plt.figure(figsize=(8, 5))
plt.plot(epochs_range, history["accuracy"], label="Training Accuracy")
plt.plot(epochs_range, history["val_accuracy"], label="Validation Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("Training and Validation Accuracy")
plt.legend()
plt.show()
````

## 16. Loss visualization

````python
plt.figure(figsize=(8, 5))
plt.plot(epochs_range, history["loss"], label="Training Loss")
plt.plot(epochs_range, history["val_loss"], label="Validation Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")
plt.legend()
plt.show()
````

## 17. Other evaluation metrics

````python
# Predict the classes for the separate test dataset
y_pred = model.predict(test_set)
predicted_categories = np.argmax(y_pred, axis=1)

# Get the actual class labels
true_categories = np.concatenate([labels.numpy() for images, labels in test_set])
````

````python
print(classification_report(
    true_categories,
    predicted_categories,
    labels=np.arange(num_classes),
    target_names=class_name,
    zero_division=0
))
````

## 18. Confusion matrix visualization

````python
cm = confusion_matrix(
    true_categories,
    predicted_categories,
    labels=np.arange(num_classes)
)

plt.figure(figsize=(18, 15))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_name,
    yticklabels=class_name
)
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("Plant Disease Classification Confusion Matrix")
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
````

## 19. Predicting a new leaf image

````python
def predict_disease(image_path):
    image = tf.keras.utils.load_img(image_path, target_size=IMG_SIZE)
    image_array = tf.keras.utils.img_to_array(image)
    image_array = np.expand_dims(image_array, axis=0)

    prediction = model.predict(image_array, verbose=0)
    predicted_index = np.argmax(prediction[0])

    print("Predicted disease:", class_name[predicted_index])
    print("Confidence:", round(float(prediction[0][predicted_index]) * 100, 2), "%")

# Example:
# predict_disease("sample_leaf.jpg")
````

## 20. Important checks if accuracy is still low

- Confirm that `train`, `valid`, and `test` contain the correct images in the correct class folders.
- Check the displayed sample images and their labels.
- Make sure the dataset paths point to the correct folders.
- The notebook checks that class folder names match across all three datasets.
- If both training and validation accuracy remain near random chance, investigate incorrect labels, empty/wrong folders, or image loading issues.
- If training accuracy is high but validation accuracy is low, the model may be overfitting or the datasets may differ.
- Do not use the test set to tune the model. Use it only for final evaluation.

**Note:** No model can guarantee a fixed accuracy. The actual result depends on the dataset, image quality, class balance, and label correctness.
